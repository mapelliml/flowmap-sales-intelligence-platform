```python
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
from pathlib import Path

# Archivo origen
csv_file = "C:\\Users\\34642\\Desktop\\FLOWMAP ANALYTICS\\segundo proyecto\\POWER BI\\data\\raw\\train.csv"

# Archivo destino
parquet_file = "C:\\Users\\34642\\Desktop\\FLOWMAP ANALYTICS\\segundo proyecto\\POWER BI\\data_processed\\train.parquet"

# Tamaño de chunk
chunksize = 500_000

# Eliminar parquet anterior si existe
if Path(parquet_file).exists():
    Path(parquet_file).unlink()

writer = None

print("Iniciando conversión...")

for i, chunk in enumerate(
    pd.read_csv(
        csv_file,
        chunksize=chunksize
    )
):

    print(f"Procesando chunk {i+1}")

    # Reducir memoria
    chunk["store_nbr"] = chunk["store_nbr"].astype("int16")
    chunk["item_nbr"] = chunk["item_nbr"].astype("int32")

    if "onpromotion" in chunk.columns:
        chunk["onpromotion"] = chunk["onpromotion"].astype("bool")

    chunk["unit_sales"] = chunk["unit_sales"].astype("float32")

    table = pa.Table.from_pandas(
        chunk,
        preserve_index=False
    )

    if writer is None:
        writer = pq.ParquetWriter(
            parquet_file,
            table.schema,
            compression="snappy"
        )

    writer.write_table(table)

if writer:
    writer.close()

print("Conversión terminada.")
print(f"Archivo generado: {parquet_file}")
```
